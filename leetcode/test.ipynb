```python
height=[1,8,6,2,5,4,8,3,7]
vol=[]

for i in range(len(height)):
    for j in range(i+1,len(height)):
        if (height[i]<height[j]):
            v=height[i]*(j-i)
            vol.append(v)
        elif(height[i]>height[j]):
            v=height[j]*(j-i)
            vol.append(v)

print(vol)
```

```python
height = [1,2,6,2,5,4,8,3,7]
vol = []

for i in range(len(height)):
    for j in range(i + 1, len(height)):
        area = min(height[i], height[j]) * (j - i)
        vol.append(area)

print(vol)
print("Maximum water:", max(vol))
```
